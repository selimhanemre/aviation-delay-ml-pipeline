```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import mlflow.xgboost

print("1. Loading the Weighted XGBoost Model from MLflow...")
# IMPORTANT: Replace this with the exact Run ID from your MLflow UI 
# Look for the run named "LIT_Flight_Delay_XGBoost_Weighted"
RUN_ID = "runs:/7d54c61ff67149c4b2def551290278b6/xgboost-model" 
loaded_model = mlflow.xgboost.load_model(RUN_ID)

print("2. Extracting Feature Importance...")
# Get the underlying XGBoost booster object
booster = loaded_model.get_booster()

# Extract the "Gain" (The average improvement in accuracy brought by a feature)
importance_dict = booster.get_score(importance_type='gain')

# Convert to a Pandas DataFrame for easy plotting
importance_df = pd.DataFrame({
    'Feature': list(importance_dict.keys()),
    'Importance': list(importance_dict.values())
})

# Sort from most important to least important
importance_df = importance_df.sort_values(by='Importance', ascending=False).reset_index(drop=True)

print("3. Generating Visualization...")
plt.figure(figsize=(12, 8))
sns.barplot(
    x='Importance', 
    y='Feature', 
    data=importance_df, 
    palette='viridis'
)

plt.title('What Causes Flight Delays out of Little Rock? (XGBoost Feature Importance)', fontsize=16, fontweight='bold')
plt.xlabel('Relative Importance (Information Gain)', fontsize=12)
plt.ylabel('Flight & Weather Features', fontsize=12)
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.tight_layout()

# Display the plot in the Databricks notebook
plt.show()

# Print the top 5 drivers to the console
print("\nTop 5 Drivers of Flight Disruptions:")
for i in range(5):
    print(f"{i+1}. {importance_df['Feature'].iloc[i]}")
```
