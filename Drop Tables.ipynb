```python
from pyspark.sql.functions import concat, col, lit

silver_flights = spark.table("aviation_project.silver_historical_flights")
silver_flights = silver_flights.withColumn("flight", concat(col("airline_code"), lit("_"), col("flight_number")))
```

```python
silver_flights.groupBy("flight").count().orderBy("count", ascending=False).show(20)
```
